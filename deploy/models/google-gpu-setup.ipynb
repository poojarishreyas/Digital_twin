```python
import json, os, secrets, subprocess, sys
from pathlib import Path
work = Path('/content/twinhunter-model')
work.mkdir(parents=True, exist_ok=True)
lock = {'model_id': 'Qwen/Qwen3.8-27B', 'revision': '1d4bf0f2ff6012fd82039f2fa52739d0dd7c60c0', 'served_name': 'twinhunter-qwen38-27b-1d4bf0f2', 'license': 'Apache-2.0', 'dtype': 'bfloat16', 'quantization': None, 'context_length': 16384, 'vllm_version': '0.32.0', 'transformers_version': '5.19.0', 'gpu_memory_utilization': 0.85, 'gpu_family': 'A100', 'minimum_gpu_memory_mib': 76000, 'validation_status': 'awaiting_gpu_run'}
(work / 'model-lock.json').write_text(json.dumps(lock))
(work / 'serve_model.py').write_text('"""Run on the GPU control-plane host. No Docker socket or sandbox network access."""\nimport json\nimport os\nimport subprocess\nimport sys\nfrom pathlib import Path\nfrom importlib.metadata import version\n\nlock = json.loads(Path(__file__).with_name("model-lock.json").read_text())\nif version("vllm") != lock["vllm_version"]:\n    raise SystemExit("Install the locked vLLM version before serving")\nif version("transformers") != lock["transformers_version"]:\n    raise SystemExit("Install the locked Transformers version before serving")\ngpu = subprocess.check_output(["nvidia-smi", "--query-gpu=name,memory.total,memory.free", "--format=csv,noheader,nounits"], text=True)\nprint(gpu)\ndevices = [line.split(",") for line in gpu.strip().splitlines()]\nif len(devices) != 1 or lock["gpu_family"].lower() not in devices[0][0].lower():\n    raise SystemExit("Select a single NVIDIA A100 runtime for this locked experiment")\nif int(devices[0][1]) < lock["minimum_gpu_memory_mib"] or int(devices[0][2]) < 70000:\n    raise SystemExit("A full, mostly free A100 80GB is required for this BF16 profile")\nkey = os.environ.get("TWIN_MODEL_API_KEY")\nif not key:\n    raise SystemExit("Set TWIN_MODEL_API_KEY; do not expose an unauthenticated endpoint")\nos.environ["VLLM_API_KEY"] = key\ncommand = ["vllm", "serve", lock["model_id"], "--revision", lock["revision"], "--served-model-name", lock["served_name"],\n           "--dtype", lock["dtype"], "--max-model-len", str(lock["context_length"]), "--gpu-memory-utilization", str(lock["gpu_memory_utilization"]),\n           "--tensor-parallel-size", "1", "--max-num-seqs", "1", "--language-model-only", "--host", "127.0.0.1", "--port", "8000"]\nPath("model-runtime-manifest.json").write_text(json.dumps({"lock": lock, "gpu": gpu, "command": command}, indent=2))\nraise SystemExit(subprocess.call(command))\n')
os.environ['TWIN_MODEL_API_KEY'] = secrets.token_urlsafe(32)
log = (work / 'server.log').open('w')
server = subprocess.Popen([sys.executable, str(work / 'serve_model.py')], cwd=work, stdout=log, stderr=subprocess.STDOUT)
```

```python
# Wait up to 15 minutes for model download and startup.
import time, urllib.error, urllib.request
deadline = time.monotonic() + 900
while time.monotonic() < deadline:
    if server.poll() is not None: raise RuntimeError('Model server exited; inspect /content/twinhunter-model/server.log')
    request = urllib.request.Request('http://127.0.0.1:8000/v1/models', headers={'Authorization': 'Bearer ' + os.environ['TWIN_MODEL_API_KEY']})
    try:
        with urllib.request.urlopen(request, timeout=10) as response: models = json.load(response)
        assert any(model['id'] == lock['served_name'] for model in models['data'])
        print('Pinned model ready:', lock['served_name'])
        break
    except (urllib.error.URLError, TimeoutError): time.sleep(10)
else: raise TimeoutError('Model did not become ready within 15 minutes; inspect server.log')
```
